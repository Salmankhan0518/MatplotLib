## Subplot function

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.subplot(1,2,1)
plt.plot(x,y)
plt.title("Line chart")


plt.subplot(1,2,2)
plt.bar(x,y)
plt.title("Bar chart")

plt.tight_layout()
plt.show()
```

### Second recommended way, Proffesional way or an Object Orianted way, with the help of this we can achieved precised control of plot and afterwards we can apply and execute many functions 

```python
import matplotlib.pyplot as plt


fig, ax = plt.subplots(1,2, figsize = (10, 5))

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

ax[0].plot(x,y, color='blue') # Line chart
ax[0].grid()
ax[0].set_title('Line Plot')

ax[1].bar(x,y, color='green') # Bar chart
ax[1].grid()
ax[1].set_title('Bar Plot')

plt.suptitle('Comparison of Line and Bar Chart')

plt.tight_layout()
plt.show()
```
