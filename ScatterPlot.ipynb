## Scattered Plot

### Defination Scattered plot is the graph where each point represent one observation with two values (x and y)

```python
import matplotlib.pyplot as plt
```

```python
hours_studies = [1, 2, 3, 4, 5, 6, 7, 8]
exam_score = [50, 55, 60, 65, 70, 75, 80, 85]

plt.scatter(hours_studies, exam_score, color = 'green', marker = 'o', label = 'Students Data')

plt.xlabel('Hours of studies')
plt.ylabel('Exam Score')
plt.title('Relationship Between Study Time and Exam Score')
plt.xticks([1, 2, 3, 4, 5, 6, 7, 8], ['1Hrs', '2Hrs', '3Hrs', '4Hrs', '5Hrs', '6Hrs', '7Hrs', '8Hrs'])
plt.legend()
plt.grid(True)
plt.show()
```

### Comperision with Multiple Classes

```python
plt.scatter([1,2,3], [50, 55, 60], color = 'blue', marker = 'o', label = 'Class A')
plt.scatter([1,2,3], [45, 50, 52], color = 'orange', marker = 'o', label = 'Class B')

plt.xlabel('Hours of studies')
plt.ylabel('Exam Score')
plt.title('Comperision of two Classes')
plt.xticks([1, 2, 3], ['1Hrs', '2Hrs', '3Hrs'])
plt.legend()
plt.grid(True)
plt.show()
```
