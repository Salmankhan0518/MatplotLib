## Introduction to Histogram

### Height of bar = No. of students in that range

#### Histogram use for show the distribution of data by rangese (bins)


###  The histogram group the continuous data which are into bins (intervals) and show the frequency of each data in each range.

```python
import matplotlib.pyplot as plt
```

```python
score = [45, 67, 56, 68, 77, 47, 78, 90, 88, 44, 33, 54, 85, 75, 92, 94, 95, 89, 81, 61, 53, 76, 99]
plt.hist(score, bins=5, color='purple', edgecolor = 'black')

plt.title('Score Distribution os Students')
plt.xlabel('Score range')
plt.ylabel('No. of students')
plt.grid()
plt.show()
```
